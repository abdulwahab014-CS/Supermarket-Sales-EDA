PART 1 – DATA IMPORT & UNDERSTANDING

```python
import pandas as pd
import numpy as np

# Load the dataset using Pandas
df = pd.read_csv('supermarket_sales.csv')
```

```python

# Display the first 5 records
print(" FIRST 5 RECORDS ")
display(df.head())

# Display the last 5 records
print(" LAST 5 RECORDS ")
display(df.tail())
```

```python

# Check dataset shape (Rows, Columns)
print(" DATASET SHAPE ")
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
```

```python
# Display column names
print(" COLUMN NAMES ")
print(df.columns.tolist())
```

```python
# Display column data types
print(" DATA TYPES ")
print(df.dtypes)

# Identify numerical and categorical columns dynamically
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

print("\n NUMERICAL COLUMNS ")
print(numerical_cols)

print("\n CATEGORICAL COLUMNS ")
print(categorical_cols)
```

```python
# Generate summary statistics for numerical columns
print(" NUMERICAL STATISTICAL SUMMARY ")
display(df.describe())

# Generate summary statistics for categorical columns
print(" CATEGORICAL STATISTICAL SUMMARY ")
display(df.describe(include=['object']))
```

PART 2 – DATA CLEANING

```python
# 1. Check Missing Values
print("Missing Values per Column:")
print(df.isnull().sum())
```

```python
# 2. Check and Remove Duplicate Records
duplicate_count = df.duplicated().sum()
print(f"\nDuplicate Records Found: {duplicate_count}")
df.drop_duplicates(inplace=True)
```

```python
# 3. Identify and Fix Incorrect Data Types
print("\nData Types Before Fix:")
print(df[['Date', 'Time']].dtypes)

# Convert string Date and Time to standard Datetime
df['Date'] = pd.to_datetime(df['Date'])
df['Datetime'] = pd.to_datetime(df['Date'].astype(str) + ' ' + df['Time'])

print("\nData Types After Fix:")
print(df[['Date', 'Datetime']].dtypes)
```

```python
# 4. Check Invalid or Inconsistent Values
print("\nValue Range & Consistency Audits:")
print(f"Unit Price Range: ${df['Unit price'].min()} to ${df['Unit price'].max()}")
print(f"Quantity Range: {df['Quantity'].min()} to {df['Quantity'].max()} units")
print(f"Rating Range: {df['Rating'].min()} to {df['Rating'].max()} stars")
print(f"Equation Check (Total = cogs + Tax 5%): {np.isclose(df['Total'], df['cogs'] + df['Tax 5%']).all()}")
```

```python
# 5. Verify Dataset After Cleaning
print("\nDataset Verification:")
print(f"Cleaned Dataset Shape: {df.shape}")
print(f"Total Remaining Missing Values: {df.isnull().sum().sum()}")
```

PART 3 – FEATURE ANALYSIS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Create folder for plots if not present
os.makedirs('graphs', exist_ok=True)

# 1. Numerical Feature Metrics
num_summary = pd.DataFrame({
    'Min': df[numerical_cols].min(),
    'Max': df[numerical_cols].max(),
    'Mean': df[numerical_cols].mean(),
    'Median': df[numerical_cols].median(),
    'Std Dev': df[numerical_cols].std()
})

print("Numerical Feature Summaries:")
display(num_summary)

```

```python
# 2. Categorical Feature Summaries
print("\nCategorical Feature Summaries:")
for col in ['Product line', 'Branch', 'City', 'Customer type', 'Gender', 'Payment']:
    top_cat = df[col].mode()[0]
    count = df[col].value_counts()[top_cat]
    print(f"{col}: {df[col].nunique()} unique categories | Top: '{top_cat}' ({count} transactions)")
```

```python
# 3. Outlier Analysis Visualization
plt.figure(figsize=(9, 4.5))
sns.boxplot(data=df[['Unit price', 'Quantity', 'Tax 5%', 'gross income', 'Rating']], palette='Set3')
plt.title("Distribution and Outliers of Numerical Features")
plt.tight_layout()
plt.savefig('graphs/feature_outliers.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# Separate Boxplot for Total Sales Outliers
plt.figure(figsize=(9, 3.5))
sns.boxplot(data=df, x='Total', color='skyblue')
plt.title("Distribution and Outliers of Total Sales Amount ($)")
plt.tight_layout()
plt.savefig('graphs/total_sales_outliers.png', dpi=200, bbox_inches='tight')
plt.show()
```

PART 4 – CORRELATION ANALYSIS

```python
# Select suitable numerical features
numerical_cols = ['Unit price', 'Quantity', 'Tax 5%', 'Total', 'cogs', 'gross income', 'Rating']
corr_matrix = df[numerical_cols].corr()

# Ensure target folder exists
os.makedirs('graphs', exist_ok=True)

# Generate Correlation Heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix Heatmap")
plt.tight_layout()
plt.savefig('graphs/heatmap_correlation.png', dpi=200, bbox_inches='tight')
plt.show()
```

PART 5 – DATA VISUALIZATION

```python
# 1. Price Distribution Histogram (KDE Curve)
plt.figure(figsize=(9, 5))
sns.histplot(df['Unit price'], kde=True, color='teal', bins=20)
plt.title('1. Price Distribution Histogram with KDE Curve')
plt.xlabel('Unit Price ($)')
plt.ylabel('Frequency')
plt.savefig('graphs/01_price_histogram.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 2. Average Price Bar Chart
plt.figure(figsize=(10, 5))
df.groupby('Product line')['Unit price'].mean().sort_values(ascending=False).plot(kind='bar', color='darkslategrey')
plt.title('2. Average Unit Price across Product Categories ($)')
plt.xlabel('Product Line')
plt.ylabel('Mean Unit Price ($)')
plt.xticks(rotation=30, ha='right')
plt.savefig('graphs/02_avg_price_barchart.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 3. Category Count Plot
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Product line', palette='Set2', order=df['Product line'].value_counts().index)
plt.title('3. Category Inventory Volume (Transaction Count Plot)')
plt.xlabel('Product Line')
plt.ylabel('Product / Order Volume')
plt.xticks(rotation=30, ha='right')
plt.savefig('graphs/03_category_countplot.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 3. Category Count Plot
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Product line', palette='Set2', order=df['Product line'].value_counts().index)
plt.title('3. Category Inventory Volume (Transaction Count Plot)')
plt.xlabel('Product Line')
plt.ylabel('Product / Order Volume')
plt.xticks(rotation=30, ha='right')
plt.savefig('graphs/03_category_countplot.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 5. Price vs Rating Count (Quantity Engagement) Scatter Plot
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Unit price', y='Quantity', hue='Customer type', alpha=0.7, s=60)
plt.title('5. Price vs. Customer Engagement (Quantity) Scatter Plot')
plt.xlabel('Unit Price ($)')
plt.ylabel('Quantity Purchased')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.savefig('graphs/05_price_vs_engagement_scatter.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python


# 6. Rolling Average Price Line Chart
df_sorted = df.sort_values(by='Unit price').reset_index(drop=True)

plt.figure(figsize=(10, 5))
df_sorted['Rolling_Avg_Price'] = df_sorted['Unit price'].rolling(window=50).mean()
plt.plot(df_sorted['Unit price'], alpha=0.3, color='gray', label='Raw Unit Price')
plt.plot(df_sorted['Rolling_Avg_Price'], color='crimson', linewidth=2.0, label='50-Period Rolling Average')
plt.title('6. Rolling Average Price Line Chart Across Sorted Catalog Items')
plt.xlabel('Sorted Catalog Index')
plt.ylabel('Unit Price ($)')
plt.legend(loc='upper left')
plt.savefig('graphs/06_rolling_price_linechart.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 7. Price Outlier Box Plot
plt.figure(figsize=(9, 4.5))
sns.boxplot(data=df, x='Total', palette='Blues')
plt.title('7. Price Outlier Box Plot (IQR Method)')
plt.xlabel('Total Sales Spend ($)')
plt.savefig('graphs/07_price_outlier_boxplot.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 8. Category Price Violin Plot
plt.figure(figsize=(10, 5))
sns.violinplot(data=df, x='Product line', y='Unit price', palette='muted', cut=0)
plt.title('8. Category Price Density and Distribution (Violin Plot)')
plt.xlabel('Product Line')
plt.ylabel('Unit Price ($)')
plt.xticks(rotation=30, ha='right')
plt.savefig('graphs/08_category_price_violinplot.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 9. Correlation Matrix Heatmap
plt.figure(figsize=(8, 5))
num_cols = ['Unit price', 'Quantity', 'Tax 5%', 'Total', 'cogs', 'gross income', 'Rating']
sns.heatmap(df[num_cols].corr(), annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title('9. Annotated Correlation Matrix Heatmap')
plt.savefig('graphs/09_correlation_heatmap.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 10. Multi-Variable Pair Plot
pair_cols = ['Unit price', 'Quantity', 'Rating', 'Total']
pair_grid = sns.pairplot(df[pair_cols], corner=True, diag_kind='kde')
pair_grid.fig.suptitle('10. Multi-Variable Pair Plot (Scatter & KDE Curves)', y=1.05, fontsize=14)
pair_grid.savefig('graphs/10_multivariable_pairplot.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 11. Rating Density Distribution Plot (KDE)
plt.figure(figsize=(9, 5))
sns.kdeplot(data=df, x='Rating', fill=True, color='purple', alpha=0.4)
plt.title('11. Rating Density Distribution Plot (KDE Curve)')
plt.xlabel('Customer Rating Score (1-10)')
plt.ylabel('Density')
plt.savefig('graphs/11_rating_kde_distribution.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# 12. Cumulative Price Area Chart
plt.figure(figsize=(10, 5))
df_sorted['Cumulative_Price'] = df_sorted['Unit price'].cumsum()
plt.fill_between(range(len(df_sorted)), df_sorted['Cumulative_Price'], color='cornflowerblue', alpha=0.5)
plt.plot(range(len(df_sorted)), df_sorted['Cumulative_Price'], color='blue', linewidth=1.5)
plt.title('12. Cumulative Price Growth Area Chart Across Catalog')
plt.xlabel('Sorted Catalog Index')
plt.ylabel('Cumulative Price ($)')
plt.savefig('graphs/12_cumulative_price_areachart.png', dpi=200, bbox_inches='tight')
plt.show()
```
